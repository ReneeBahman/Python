##Sorting with dogs example: 


`pd.read_csv("data/dogs.csv")` ## imports dogs.csv if it would exist 

`dogs = pd.read_csv("data/dogs.csv")`  #assigns name dogs to the dataset 

`dogs.sort_values("weight_kg")` # sorts dogs arranging the list from weight perspective ascending - the smallest weight dog first and then heaviest 

`dogs.sort_values("weight_kg", ascending=False)`   # sorts dogs arranging the list from weight perspective descengind - now heaviest dog is first 

`dogs.sort_values(["weight_kg", "height_cm"])`  # sorts dogs arranging the the list with weight and hight 

`dogs.sort_values(["weight_kg", "height_cm"], ascending=[True, False])`  # sorts dogs arranging the the list with weight and hight with specification, which value is ascending and which is descending
